```python
import numpy as np
import pandas as pd
import torch
from torch import nn
```

```python
class FeaturePyramidNetwork(nn.Module):
    def __init__(self, c2, c3, c4, c5, out_channels):
        super().__init__()
        self.c2_lateral = nn.Conv2d(c2, out_channels, kernel_size=1)
        self.c3_lateral = nn.Conv2d(c3, out_channels, kernel_size=1)
        self.c4_lateral = nn.Conv2d(c4, out_channels, kernel_size=1)
        self.c5_lateral = nn.Conv2d(c5, out_channels, kernel_size=1)
        self.smoothing = nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1)



    def forward(self, x):
        c2, c3, c4, c5 = x
        p5 = self.c5_lateral(c5)
        p4 = self.c4_lateral(c4) + nn.functional.upsample_nearest(p5, scale_factor=2)
        p3 = self.c3_lateral(c3) + nn.functional.upsample_nearest(p4, scale_factor=2)
        p2 = self.c2_lateral(c2) + nn.functional.upsample_nearest(p3, scale_factor=2)
        p5 = self.smoothing(p5)
        p4 = self.smoothing(p4)
        p3 = self.smoothing(p3)
        p2 = self.smoothing(p2)
        return p2, p3, p4, p5
```
