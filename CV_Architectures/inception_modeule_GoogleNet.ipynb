```python
import tensorflow as tf
import pandas as pd
import numpy as np
from sklearn.datasets import load_sample_image
```

```python
def build_inception_module(input_, f1, f2, f3, f4, f5, f6):
  tower1 = tf.keras.layers.Conv2D(filters=f1, kernel_size=1, strides=(1, 1), padding='same', activation='relu')(input_)

  tower2 = tf.keras.layers.Conv2D(filters=f5, kernel_size=1, strides=(1, 1), padding='same', activation='relu')(input_)
  tower2 = tf.keras.layers.Conv2D(filters=f2, kernel_size=3, strides=(1, 1), padding='same', activation='relu')(tower2)

  tower3 = tf.keras.layers.Conv2D(filters=f6, kernel_size=1, strides=(1, 1), padding='same', activation='relu')(input_)
  tower3 = tf.keras.layers.Conv2D(filters=f3, kernel_size=5, strides=(1, 1), padding='same', activation='relu')(tower3)

  tower4 = tf.keras.layers.MaxPool2D(pool_size=(3, 3), strides=(1, 1), padding='same')(input_)
  tower4 = tf.keras.layers.Conv2D(filters=f4, kernel_size=1, strides=(1, 1), padding='same', activation='relu')(tower4)

  output = tf.concat(values=[tower1, tower2, tower3, tower4], axis=3)
  return output
```
