```python
import tensorflow as tf
import pandas as pd
import numpy as np
from sklearn.datasets import load_sample_image
```

```python
mnist = tf.keras.datasets.mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
rows, cols = 28, 28
input_shape = (rows, cols, 1)

x_train = x_train.reshape(x_train.shape[0], rows, cols, 1)
x_test  = x_test.reshape(x_test.shape[0], rows, cols, 1)

x_train = x_train.astype('float32')
x_test  = x_test.astype('float32')

x_train /= 255
x_test  /= 255
```

```python
y_train = tf.keras.utils.to_categorical(y_train, 10)
y_test  = tf.keras.utils.to_categorical(y_test, 10)
```

```python
def build_le_net(input_shape):
  model = tf.keras.Sequential()

  # C1
  model.add(tf.keras.layers.Conv2D(6, (5, 5), strides=1, padding='same', activation='tanh', input_shape=input_shape))

  # S2
  model.add(tf.keras.layers.AveragePooling2D(pool_size=(2, 2), strides=2, padding='valid'))

  # C3
  model.add(tf.keras.layers.Conv2D(16, (5, 5), strides=1, padding='valid', activation='tanh'))

  # S4
  model.add(tf.keras.layers.AveragePooling2D(pool_size=(2, 2), strides=2, padding='valid'))

  # C5
  model.add(tf.keras.layers.Conv2D(120, (5, 5), strides=1, padding='valid', activation='tanh'))

  # F6
  model.add(tf.keras.layers.Flatten())
  model.add(tf.keras.layers.Dense(84, activation='tanh'))

  # Out
  model.add(tf.keras.layers.Dense(10, activation='softmax'))

  return model
```

```python
lenet = build_le_net(input_shape)

lenet.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

lenet.fit(x_train, y_train, batch_size=128, epochs=20, verbose=1)
```

```python
loss, accuracy = lenet.evaluate(x_test, y_test, verbose=0)
print('Test loss:', loss)
print('Test accuracy:', accuracy)
```
